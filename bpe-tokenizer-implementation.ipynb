```python
from collections import defaultdict

word_freqs = defaultdict(int) # int = 0 set by default 

corpus = [
    "Lionel Messi was born on June 24, 1987 in Rosario, Argentina.",
    "He is widely regarded as one of the greatest footballers of all time.",
    "Messi began playing football at a young age and joined the Newell's Old Boys youth team.",
    "At age 13, he moved to Barcelona to join FC Barcelona’s famed La Masia academy.",
    "Messi made his competitive debut for Barcelona in 2004 at the age of 17.",
    "Over the next 17 years, Messi became the club's all-time top scorer and won numerous trophies.",
    "He won 10 La Liga titles, 7 Copa del Rey trophies, and 4 UEFA Champions League titles with Barcelona.",
    "Messi has won the Ballon d'Or award a record number of times, recognizing him as the world's best player.",
    "In August 2021, Messi signed with Paris Saint-Germain (PSG) after leaving Barcelona due to financial issues at the club.",
    "At PSG, Messi teamed up with other stars like Neymar Jr. and Kylian Mbappe.",
    "Despite a challenging first season in Paris, Messi continued to showcase his playmaking and goal-scoring abilities.",
    "Internationally, Messi helped Argentina win the Copa America in 2021, his first major international trophy.",
    "In 2022, Messi led Argentina to victory at the FIFA World Cup in Qatar, cementing his legacy.",
    "His performances throughout the tournament earned him the Golden Ball as the best player of the World Cup.",
    "Messi is known for his incredible dribbling, vision, passing, and goal-scoring ability.",
    "Fans and analysts often compare Messi to Cristiano Ronaldo in debates over who is the greatest footballer of their era.",
    "Off the field, Messi is involved in various charitable activities and runs the Leo Messi Foundation, focusing on health and education.",
    "Messi is also a global sports icon, attracting sponsorships and endorsements from major brands.",
    "Throughout his career, Messi has inspired millions of fans around the world with his humility and dedication.",
    "Even in the later stages of his career, Messi continues to deliver magical moments on the pitch."
]

for text in corpus:
    words_with_offsets = tokenizer.backend_tokenizer.pre_tokenizer.pre_tokenize_str(text)
    new_words = [word for word, offset in words_with_offsets]
    for word in new_words:
        word_freqs[word] += 1

print(word_freqs)
```

```python
alphabet = []

for word in word_freqs.keys():
    for letter in word:
        if letter not in alphabet:
            alphabet.append(letter)
alphabet.sort()

print(alphabet)
```

```python
vocab = ["<|endoftext|>"] + alphabet.copy()
print(vocab)
```

```python
splits = {word: [c for c in word] for word in word_freqs.keys()}
splits
```

```python
def compute_pair_freqs(splits):
    pair_freqs = defaultdict(int)
    for word, freq in word_freqs.items():
        split = splits[word]
        if len(split) == 1: # this is single letter
            continue
        for i in range(len(split) - 1):
            pair = (split[i], split[i+1])
            pair_freqs[pair] += freq
    return pair_freqs
```

```python
pair_freqs = compute_pair_freqs(splits)
for i, key in enumerate(pair_freqs.keys()):
    print(f"{key}: {pair_freqs[key]}")
    if i >= 5:
        break
```

```python
best_pair = ""
max_freq = None

for pair, freq in pair_freqs.items():
    if max_freq is None or max_freq < freq:
        best_pair = pair
        max_freq = freq

print(best_pair, max_freq)
```

```python
def merge_pair(a, b, splits):
    for word in word_freqs:
        split = splits[word]
        if len(split) == 1:
            continue

        i = 0
        while i < len(split) - 1:
            if split[i] == a and split[i + 1] == b:
                split = split[:i] + [a + b] + split[i + 2 :]
            else:
                i += 1
        splits[word] = split
    return splits
```

```python
splits = merge_pair("Ġ", "t", splits)
print(splits["Ġthroughout"])
```

```python
merges = {("Ġ", "t"): "Ġt"}
vocab.append("Ġt")
```

```python
vocab_size = 1000

while len(vocab) < vocab_size:
    pair_freqs = compute_pair_freqs(splits)

    best_pair = None
    max_freq = None
    for pair, freq in pair_freqs.items():
        if max_freq is None or freq > max_freq:
            best_pair = pair
            max_freq = freq

    # If no more pairs to merge, exit loop
    if best_pair is None:
        break

    # Unpack best_pair and merge
    a, b = best_pair
    splits = merge_pair(a, b, splits)

    # Record the merge
    merges[best_pair] = a + b
    vocab.append(a + b)
```
